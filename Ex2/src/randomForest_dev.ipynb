```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#import regressiontree from sklearn
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

from randomForest import RandomForest
from cross_validation import custom_cross_validate
```

```python
# Dummy Data
def f(x):
    mu, sigma = 0, 1.5
    return -x**2 + x + 5 + np.random.normal(mu, sigma, 1)

num_points = 300
np.random.seed(1)
    
x = np.random.uniform(-2, 10, num_points)
y = np.array( [f(i) for i in x] )

df = pd.DataFrame({'x': x, 'y': y.ravel()})


plt.scatter(x, y, s = 5)
```

```python
#train test split
df_train, df_test = train_test_split(df, test_size = 0.2)
```

```python
forest = RandomForest(n_estimators=100, max_depth=7, min_samples_split=5, max_features='sqrt', n_jobs=2)
```

```python
forest.fit(df_train[['x']], df_train['y'])
```
